```python
import nbformat as nbf

# 1. Crear un nuevo objeto de tipo Notebook
ntbk = nbf.v4.new_notebook()

# 2. Leer tu archivo .txt con el código
with open("tu_archivo.txt", "r", encoding="utf-8") as f:
    codigo_texto = f.read()

# 3. Añadir el contenido como una celda de código en el notebook
ntbk['cells'] = [nbf.v4.new_code_cell(codigo_texto)]

# 4. Guardar el resultado como archivo .ipynb
with open("notebook_generado.ipynb", "w", encoding="utf-8") as f:
    nbf.write(ntbk, f)

print("¡Conversión exitosa!")
```
