```python
import sys
import os
import random
import time
import networkx as nx
import osmnx as ox

# Añadir la ruta raíz del proyecto para poder acceder a la carpeta src
sys.path.append(os.path.abspath(os.path.join('..')))
from src import fase2

print("Librerías importadas correctamente para la Fase 2.")

# Cargar el grafo normal y el grafo proyectado en metros para las heurísticas
G, G_proyectado = fase2.cargar_grafo_urbano()

# Seleccionar un origen aleatorio y destinos alcanzables
nodos = list(G.nodes)
random.seed(42)  # Mantener reproducibilidad
origen = random.choice(nodos)

alcanzables = [n for n in nx.descendants(G, origen)]
alcanzables = [nodo for nodo in alcanzables if nodo != origen]

destinos = random.sample(alcanzables, min(5, len(alcanzables)))
print(f"Origen seleccionado: {origen}")
print(f"Destinos de prueba seleccionados: {len(destinos)}")

# Ejecutar los experimentos comparativos de rendimiento (A*, Greedy y UCS)
resultados = fase2.ejecutar_experimentos(G, G_proyectado, origen, destinos)

# Mostrar la tabla de resultados detallada
fase2.mostrar_tabla_resultados(resultados)

# Seleccionar una muestra de nodos para verificar la admisibilidad de h1, h2 y h3
nodos_admisibilidad = random.sample(alcanzables, min(30, len(alcanzables)))
resultados_admisibilidad = fase2.verificar_admisibilidad(G, G_proyectado, destinos[0], nodos_admisibilidad)

# Mostrar el reporte de admisibilidad
fase2.mostrar_admisibilidad(resultados_admisibilidad)

# Generar y guardar el mapa interactivo comparando A* y Greedy
fase2.visualizar_rutas(G, G_proyectado, origen, destinos[0])
```
